```python
##write function that takes tuple as input and returns minimum, maximum and sum of the elements
def tuple_operations(tup):
    minimum = min(tup)
    maximum = max(tup)
    total = sum(tup)

    return minimum, maximum, total

tup = (10, 5, 20, 3, 15)

minimum, maximum, total = tuple_operations(tup)

print("Minimum:", minimum)
print("Maximum:", maximum)
print("Sum:", total)
```
