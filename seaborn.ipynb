```python
import pandas as pd
tips = pd.read_csv('tips.csv')
tips
```

```python
##create a scatter plot
import matplotlib.pyplot as plt
import seaborn as sns
sns.scatterplot(x='total_bill',y='tip',data=tips)
plt.title('Scatter plot')
plt.show()
```

```python
##line plot
sns.lineplot(x='size',y='total_bill',data=tips)
```

```python
##bar plot
sns.barplot(x='day',y='total_bill',data=tips)
```

```python
##box plot
sns.boxplot(x='day',y='total_bill',data=tips)
```

```python
##violin plot
sns.violinplot(x='day',y='total_bill',data=tips)
```

```python
##histograms
sns.histplot(tips['total_bill'],bins=20,kde=False)
```

```python
##kde plot
sns.kdeplot(tips['total_bill'],fill=False)
```

```python
##pair plot
sns.pairplot(tips)
```

```python
sales_df = pd.read_csv('sales_data.csv', header=1, index_col=0)
print(sales_df.columns.tolist())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10, 6))
sns.barplot(x='Product Category', y='Total Revenue', data=sales_df,
            estimator='sum', errorbar=None)
plt.xlabel('Product Category')
plt.ylabel('Total Revenue')
plt.title('Total sales by product')
plt.xticks(rotation=45)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10, 6))
sns.barplot(x='Region', y='Total Revenue', data=sales_df,
            estimator='sum', errorbar=None)
plt.xlabel('Product Category')
plt.ylabel('Total Revenue')
plt.title('Total sales by region')
plt.xticks(rotation=45)
plt.show()
```
