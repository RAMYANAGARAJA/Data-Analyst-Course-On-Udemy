```python
##create a pandas dataframe with 2 columns 'Category' and 'Value'. Fill the 'Category' column with random categories. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' of each category
df=pd.DataFrame({'Category':np.random.choice(['A','B','C'],size=10),'Value':np.random.randint(1,100,size=10)})
print(df)
grouped=df.groupby('Category')['Value'].agg(['sum','mean'])
print(grouped)
```

```python
##create a pandas dataframe with 3 columns: 'Product',Category',and 'Sales'.Fill the dataframe with random values. Group the dataframe by category and calculate the sales for each category
import numpy as np
import pandas as pd
df=pd.DataFrame({'Category':np.random.choice(['A','B','C'],size=10),'Product':np.random.choice(['P1','P2','P3'],size=10),'Sales':np.random.randint(100,200,size=10)})
print(df)
saleseach=df.groupby('Category')['Sales'].sum()
print(saleseach)
```

```python
##create two pandas dataframes with a common column. Merge the dataframes using the common column
import numpy as np
import pandas as pd
df1=pd.DataFrame({'key':['A','B','C','D'],'Product':['TV','Fridge','Oven','Phone']})
df2=pd.DataFrame({'key':['A','B','C','D'],'Count':[6,15,13,21]})
print(df1)
print(df2)
result=pd.merge(df1,df2,on='key')
print(result)
```

```python
##create two dataframes along with different columns. Concatenate the dataframes along with rows and columns
import numpy as np
import pandas as pd
df1=pd.DataFrame(({'A':np.random.randint(1,5,size=5),'B':np.random.randint(1,5,size=5)}))
df2=pd.DataFrame(({'C':np.random.randint(1,5,size=5),'D':np.random.randint(1,5,size=5)}))
print(df1)
print(df2)
concatrows=pd.concat([df1, df2],axis=0)
print(concatrows)
concatcols=pd.concat([df1, df2],axis=1)
print(concatcols)
```

```python
##create a pandas dataframe with a date time index and one column filled with random integers. Resample the dataframe to compute the monthly mean of the values
import pandas as pd
import numpy as np
date_rng=pd.date_range(start='2025-01-01',end='2025-12-31',freq='D')
df=pd.DataFrame(date_rng,columns=['date'])
df['data']=np.random.randint(1,10,size=(len(date_rng)))
df.set_index('date',inplace=True)
monthlymean=df.resample('ME').mean()
print(monthlymean)
```

```python
##create a pandas dataframe with a datetime index ranging from '2021-01-01' to '2021-12-31' and one column filled with random integers. Compute the rolling mean with a window of 7 days
import numpy as np
import pandas as pd
date_rng=pd.date_range(start='2021-01-01',end='2021-12-31',freq='D')
df=pd.DataFrame(date_rng,columns=['date'])
df['data']=np.random.randint(1,100,size=(len(date_rng)))
df.set_index('date',inplace=True)
sevenmean=df.rolling(window=7).mean()
print(sevenmean)
```

```python
##create a pandas dataframe with a multi index (hierarchical index). Perform some basic indexing and slicing operations on Multi Index Data Frame
arrays=[['A','A','B','B'],['one','two','one','two']]
index=pd.MultiIndex.from_arrays(arrays,names=('Category','Subcategory'))
df=pd.DataFrame(np.random.randint(1,100,size=(4,3)),index=index,columns=['Value1','Value2','Value3'])
print(df)
print('Indexing at category A')
print(df.loc['A'])
print("Slicing at category 'B'and subcategory 'two':")
print(df.loc['B','two'])
```

```python
##Create a pandas dataframe with multiindex consisting of 'Category' and'Subcategory'.Fill the dataframe with random data and compute the sum of values for each 'category' and 'sub category'
import numpy as np
import pandas as pd
arrays=[['A','A','B','B'],['one','two','one','two']]
index=pd.MultiIndex.from_arrays(arrays,names=('Category','Subcategory'))
df=pd.DataFrame(np.random.randint(1,100,size=(4,3)),index=index,columns=['Value1','Value2','Value3'])
print(df)
categorywise=df.groupby('Category','Subcategory').sum()
print(categorywise)
```

```python
##create a pandas DataFrame with columns 'Date', 'Category' and 'Value'.Create a pivot table the sum of 'Value' for each 'category' by 'Date'
date_rng=pd.date_range(start='2025-01-01',end='2025-12-31',freq='D')
df=pd.DataFrame({'Date':np.random.choice(date_rng,size=20),'Category':np.random.choice(['A','B','C'],size=20),'Value':np.random.randint(1,100,size=20)})
print(df)
pivot_table=df.pivot_table(values='Value',index='Date',columns='Category',aggfunc='sum')
print(pivot_table)
```

```python
import pandas as pd
import numpy as np
df = pd.DataFrame([
    {
        "Year": year,
        "Quarter": quarter,
        "Revenue": np.random.randint(1000, 10000),
    }
    for year in range(2000, 2026)
    for quarter in range(1, 5)
])
pivot = df.pivot_table(
    index="Quarter",
    columns="Year",
    values="Revenue",
    aggfunc="mean",
)
print(pivot)
```
