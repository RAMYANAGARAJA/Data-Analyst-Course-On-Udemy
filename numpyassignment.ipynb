```python
##create a numpy array of size(4,4) with values from 1 to 16. Compute the rowise and columnwise sum
import numpy as np
arr=np.random.randint(1,16,size=(4,4))
print(arr)
for i in range (0,4):
    sum=0
    for j in range (0,4):
        sum=sum+arr[i][j]
    print(f"row sum {i}=",sum)

for i in range(0,4):
    sum=0
    for j in range(0,4):
        sum=sum+arr[j][i]
    print(f"col sum {i}=",sum)
```

```python
##create a numpy array of shape (5,5) filled with random integers. Compute the mean, median, standard deviation and variance of the array
import numpy as np
arr=np.random.randint(1,100,size=(5,5))
array_mean=np.mean(arr)
print(arr)
print(f"mean={array_mean}")
array_median=np.median(arr)
print(f"median={array_median}")
array_sd=np.std(arr)
print(f"standard deviation={array_sd}")
array_variance=np.var(arr)
print(f"variance={array_variance}")
```

```python
##create a numpy array of size (3,3) having values from 1 to 9. Normalize the array
import numpy as np
arr = np.arange(1,10).reshape(3,3)
data = np.array(arr)
print(arr)
data_mean = np.mean(data)
std_dev = np.std(data)
normalized_data = (data - data_mean) / std_dev
print(normalized_data)
```

```python
##create a numpy array of shape (3,3) representing a matrix. Compute its determinant, inverse and eigen values
import numpy as np
arr=np.random.randint(1,10,size=(3,3))
determinant=np.linalg.det(arr)
print(f"determinant={determinant}")
inverse=np.linalg.inv(arr)
print(f"inverse={inverse}")
eigenvalues=np.linalg.eig(arr)
print(f"Eigen values={eigenvalues}")
```

```python
##create a structured array with fields 'name'(string),'age'(integer), 'weight'(float). Add some data and sort the array by age
data_type=[('name','U10'),('age','i4'),('weight','f4')]
data=np.array([('Alice',26,56),('Joe',21,67),('Bob',28,70.7)],dtype=data_type)
print(data)
sorted=np.sort(data,order='age')
print(sorted)
```

```python
##create a structured array with fields 'x' and 'y' (both integers). Add some data and compute the euclidean distance between each pair of points.
import numpy as np
data_type=[('x','i4'),('y','i4')]
data=np.array([(3,5),(7,4),(2,4)],dtype=data_type)
print(data)
distances = np.sqrt((data['x'][:, np.newaxis] - data['x'])**2 + (data['y'][:, np.newaxis] - data['y'])**2)
print("Euclidean distances:")
print(distances)
```

```python
##create a masked array of shape (4,4) with random integers and mask the elements greater than 10. Compute the sum of the unmasked elements
import numpy as np
import numpy.ma as ma
arr=np.random.randint(1,17,size=(4,4))
print(arr)
masked=ma.masked_greater(arr,10)
print(masked)
sum=masked.sum()
print(sum)
```

```python
##create a masked array of shape (3,3) with random itegers and mask the diagonal elements. Replace the masked elements with the mean of the unmasked elements
import numpy as np
array = np.random.randint(1, 21, size=(3, 3))
masked_array = ma.masked_array(array, mask=np.eye(3, dtype=bool))
print("Original array:")
print(array)
print("Masked array:")
print(masked_array)

# Replace the masked elements with the mean of the unmasked elements
mean_unmasked = masked_array.mean()
masked_array = masked_array.filled(mean_unmasked)
print("Modified masked array:")
print(masked_array)
```
