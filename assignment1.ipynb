```python
## write a program that asks the user to input a number until 0 is given as input, the program should output the sum of all input numbers
sum=0
while True:
    i=int(input("Enter the number:"))
    if i==0:
        break
    else:
        sum=sum+i
print(sum)
```

```python
##write a python program that asks for an input number and prints even numbers from 1 to that number using for loop
num=int(input("Enter the number:"))
for i in range(1,num+1):
    if(i%2==0):
        print(i)
```

```python
##write a python program which prints the factorial of an input number using while loop
num=int(input("Enter the number:"))
i=1
fact=1
while i<=num:
    fact=fact*i
    i=i+1
print(fact)
```

```python
##write a python program to print the sum of digits of a number given as input
num=int(input("Enter the number:"))
sum=0
while(num>0):
    digit=num%10
    sum=sum+digit
    num=num//10
print(sum)
```

```python
##write a python program that checks if a number given as an input by the user is a prime number using for loop
num=int(input("Enter the nunber:"))
for i in range(2,num):
    if num%i==0:
        break
else:
    print("prime")
```

```python
##write a python program which prints first n fibonacci numbers, where n is given as input
n=int(input("Enter n:"))
count=1
a=0
b=1
print(a)
print(b)
while(count<=n-2):
    c=a+b
    print(c)
    count=count+1
    a=b
    b=c
    
```
