```python
##odd and even function
def odd_or_even(num):
    if num%2==0:
        print("EVEN")
    else:
        print("ODD")

odd_or_even(235732)
```

```python
##default parameters
def greet(name):
    print(f"Hello {name}")

greet("Ramya")
```

```python
##positional arguments
def print_numbers(*krish):
    for number in krish:
        print(number)

print_numbers(1,2,3,4,5,6,7,"krish")
```

```python
##keyword arguments
def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")

print_details(name="krish",age=32,country="India")
```

```python
##positional and keyword arguments
def print_details(*args,**kwargs):
    for val in args:
        print(f"Positional arguments:{val}")
    for key,val in kwargs.items():
        print(f"{key}:{val}")

print_details(1,2,3,4,name="krish",age=32,country="India")
```

```python
##return statements
def multiply(a,b):
    return a*b,a,b

multiply(2,8)
```

```python
##temperature conversion
def convert_temperature(temp,unit):
    if unit =='c':
        return temp*(9/5)+32
    elif unit=='f':
        return (temp-32)*5/9
    else:
        return None

print(convert_temperature(35,'c'))
print(convert_temperature(77,'f'))
```

```python
##password strength checker
def is_strong_pass(password):
    if len(password)<8:
        return False
    if not any (char.isdigit() for char in password):
        return False
    if not any (char.islower() for char in password):
        return False
    if not any (char.isupper() for char in password):
        return False
    if not any (char in '!@#$%^&*()/<:;' for char in password):
        return False
    return True

print(is_strong_pass("weakpassword"))
print(is_strong_pass("Str0ngPwd!"))
    
```

```python
##total cost of items in a shopping cart
def calculate_total_cost(cart):
    total_cost=0
    for item in cart:
        total_cost+=item['price']*item['quantity']
    return total_cost

cart=[
    {'name':'Apple','price':0.5,'quantity':4},
    {'name':'Strawberry','price':0.3,'quantity':6},
    {'name':'Orange','price':0.7,'quantity':3}
]
total_cost=calculate_total_cost(cart)
print(total_cost)
```

```python
##palindrome questions
def is_palindrome(s):
    s=s.lower().replace(" ","")
    return s==s[::-1]

print(is_palindrome("A man a plan a canal panama"))
print(is_palindrome("Hello"))
```

```python
##factorial of a number using recursion
def factorial(n):
    if n==0:
        return 1
    else:
        return n*factorial(n-1)

print(factorial(5))
```

```python
##function to read a file and count the frequency of each word
def count_word_frequency(file_path):
    word_count={}
    with open(file_path,'r') as file:
        for line in file:
            words=line.split()
            for word in words:
                word=word.lower().strip('.,!@#$%^&*?')
                word_count[word]=word_count.get(word,0)+1

    return word_count

filepath='text.txt'
word_frequency=count_word_frequency(filepath)
print(word_frequency)
```

```python
import re

# Email validation function
def is_valid_email(email):
    pattern = r'^[a-zA-Z0-9_.+-]+@[a-zA-Z0-9_.+-]+\.[a-zA-Z0-9-.]+$'
    return re.match(pattern, email) is not None

print(is_valid_email("test@example.com"))
print(is_valid_email("invalid_email"))
```
