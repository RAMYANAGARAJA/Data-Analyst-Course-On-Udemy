```python
##read a whole file
with open('source.txt','r') as file:
    content=file.read()
    print(content)
```

```python
##read a file line by line
with open('source.txt','r') as file:
    for line in file:
        print(line.strip())
```

```python
##writing the file(overwriting)
with open('source.txt','w') as file:
    file.write('Hello world!\n')
    file.write('This is a new line\n')
```

```python
##writing the file(without overwritting)
with open('source.txt','a') as file:
    file.write('This is the start of the append operation!')
```

```python
##writing a list of lines in a file 
lines=['first line\n','second line\n','Third line\n']
with open('text.txt','w') as file:
    file.writelines(lines)
```

```python
##writing in a binary file
data=b'\x00\x01\x02\x03\x04'
with open ('example.bin','wb') as file:
    file.write(data)
```

```python
##reading a binary file
with open('example.bin','rb') as file:
    content=file.read()
    print(content)
```

```python
##reading content from a source file and copying it to a destination file
with open("source.txt",'r') as source_file:
    content=source_file.read()

with open('destination.txt','w') as destination_file:
    destination_file.write(content)
```

WORKING WITH FILE PATHS

```python
import os
cwd=os.getcwd()
print(f"current working directory is {cwd}")
```

```python
##creating a new directory
new_directory="package"
os.mkdir(new_directory)
print(f"directory {new_directory} created")
```

```python
##listing files and directories
items=os.listdir('.')
print(items)
```

```python
##joining paths
dir_name='folder'
file_name='file.txt'
file_path=os.path.join(dir_name,file_name)
print(file_path)
##getting the full path
full_path=os.path.join(os.getcwd(),dir_name,file_name)
print(full_path)
```

```python
##checking whether a file path is present
path='example1.txt' 
if os.path.exists(path):
    print(f'the path {path} exists')
else:
    print(f'the path {path} does not exist')
```

```python
##checking whether a path is a file or a directory
path='text.txt'
if(os.path.isfile(path)):
    print(f"the path {path} is a file")
elif os.path.isdir(path):
   print(f"the path {path} is a directory")
else:
    print(f"the path {path} is neither a file or a directory")
```

```python
##getting the absolute path
relative_path='example.txt'
absolute_path=os.path.abspath(relative_path)
print(absolute_path)
```
