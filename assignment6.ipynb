```python
##define a function that calculates the factorial of a given number. Test the function with different inputs
def fact(num):
    if(num==1):
        return 1
    else:
        return num*fact(num-1)

fact(5)
```

```python
###Define a decorator function that prints 'Executing function...' before executing a function and 'Function executed.' after executing it. 
## Apply this decorator to a function that takes a list of integers and returns their sum. Test the decorated function with different lists.
def decorator(func):
    def wrapper(numbers):
        print("Executing function....")

        output = func(numbers)

        print("Function executed")

        return output

    return wrapper


@decorator
def sum_list(numbers):
    answer = 0

    for i in numbers:
        answer = answer + i

    return answer


list1 = [1, 2, 3, 4, 5]
list2 = [10, 20, 30]
list3 = [2, 4, 6, 8]

print(sum_list(list1))
print(sum_list(list2))
print(sum_list(list3))
```
