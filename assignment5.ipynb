```python
##create a dictionary with first 10 positive integers as keys and their squares as values. Print the dictionary
d={i:i**2 for i in range (1,11)}
print(d)
```

```python
##print the value of the key 5 and the keys of the dictionary made 
print(f"value of the fifth key:{d[5]}")
print(f"the keys are: {list(d.keys())}")
```

```python
##add a new key-value pair(11,121) to the dictionary created and remove key value pair with key 1, print the modified dictionary
d[11]=121
d.pop(1)
print(d)
```

```python
##iterate throught the dictionary and print all the key value pairs
for key,value in d.items():
    print(f"{key}:{value}")
```

```python
##create a new dictionary containing the cubes of the first 10 positive integers using a dictionary comprehension. Print the new dictionary
d={i:i**3 for i in range (1,11)}
print(d) 
```

```python
##create two dictionaries, one with keys of first 5 positive integers and values as their squares, and other with keys of next 5 positive numbers and their squares 
##merge these into a single dictionary and print it
d1={i:i**2 for i in range(1,6)}
d2={j:j**2 for j in range(6,11)}
merged={**d1,**d2}
print(merged)
```

```python
##create a dictionary where the keys are the first 5 positive integers and the values are lists containing the first 5 multiples of the key, print the numbers
d={j:{i*j for i in range(1,6)}for j in range(1,6)}
print(d)
```

```python
##create a dictionary where the keys are the first 5 positive numbers and the values are tuples containing keys and squares. Print the dictionary
dict={i:(i,i**2)for i in range(1,6)}
print(dict)
```

```python
##create a dictionary with the first 5 positive integers as keys and their squares as values, convert the dictionary to a list of tupes and print the list
d={i:i**2 for i in range(1,6)}
list_of_tuples=list[d.items()]
print(list_of_tuples)
```

```python
##create a dictionary with the first 10 positive integers as keys and their squares as values. Create a new dictionary containing only the keys value pairs where key is even. Print the dictionary
d={i:i**2 for i in range(1,11)}
d1={k:v for k, v in d.items() if k%2==0}
print(d1)
```

```python
##create a dictionary with the first 5 positive integers as keys and their squares as values. Create a new dictionary with keys and values swapped, print the new dictionary
d={i:i**2 for i in range(1,6)}
d1={v:k for k, v in d.items()}
print(d1)
```

```python
##write a function that takes a string as input and outputs a dictionary having frequency of characters in the string
def count_freq(a):
    count_dict={}
    for char in a:
        count_dict[char]=count_dict.get(char,0)+1
    return count_dict
string="dictionary helps you understand"
print(count_freq(string))
```
