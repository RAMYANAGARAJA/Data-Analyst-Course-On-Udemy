```python
##create a list of dictionaries where each dictionary represents a student with 'keys' and 'score'
##sort the list by score in decreasing order and print the sorted list 
students=[{"name":"Ramya","score":85},
          {"name":"Anjali","score":90},
          {"name":"Sakshee","score":98},
          {"name":"Vaibhav","score":65},
          ]
students_sorted=sorted(students,key=lambda x:x["score"],reverse=True)
print(students_sorted)
```

```python
##write a function that takes a nested list as input anf flattens it into a single list. Print the original and flattened list
matrix=[]
for i in range(3):
    row=list(map(int,input("Enter row:").split()))
    matrix.append(row)

print(matrix)
flattened_list=[]
for i in range(3):
    for j in range(3):
        flattened_list.append(matrix[i][j])

print(flattened_list)
```
